eXCEPTION  HANDLING IN PYTHON


Exception handling is used to prevent a program from crashing when an error occurs.


python provides four main blocks

.try
.except
.else
.finally



try:
# code that may cause an  error
except:
# Runs if an error occurs
else:
# Run if no error occurs
finally:
# always run

```python
# 1.try Block
# Put the code that might an error inside "try"


num = int(input("Enter a number"))
print(num)
```

```python
# 2. except block:
# If an error happens in try , python jumps to except.

try:
    num = int(input("Enter a number:"))
    print(num)
except:
    print("An error occurred while taking input.")    
```

```python
# 3. else block:
#  The else block runs if no excepton occurs

try:
    num = int(input("Enter a number: "))

except:
    print("Invalid input ")
else:
    print("you entered :",num)    
```

```python
# 4. finally block:
# The finally block always excutes whether an error occurs or not.

try:
    num = int(input("Enter a number:"))
except:
    print("Invalid input")
else:
    print("you entered :",num)
finally:
    print("This block always executes")            
```

```python
# complete example

try:
    num1 = int(input("Enter 1st number:"))
    num2 = int(input("Enter 2nd number"))

    result = num1/num2
except ZeroDivisionError:
    print("Can't divide by zero") 
except ValueError:
     print("invalid input")   

else:
    print("Result is:",result)

finally:
    print("All task is completed")
```

```python
# File handling and exception handling

try:
    file = open("example.txt","r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("The file doesn't exist.")
finally:
    if "file" in locals() and not file.closed():
        file.close()
        print("file is closed")    
```
