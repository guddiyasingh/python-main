```python
def find_min(t):
    minimum = t[0]
    for i in t:
        if i < minimum:
            minimum = i
    return minimum
numbers=(5,2,9,1,3)
min_value=find_min(numbers)
print("Minimum value in the tuple:", min_value)
```

```python
def find_max(t):
    maximum = t[0]
    for i in t:
        if i > maximum:
            maximum = i
    return maximum
numbers=(5,2,9,1,3)
max_value=find_max(numbers)
print("Maximum value in the tuple:", max_value)
```
