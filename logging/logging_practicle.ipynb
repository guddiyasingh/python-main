```python
# 3.Practicle Example: Calculator

import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s-%(levelname)s-%(message)s"

)
def divide(a,b):
    logging.info(f"Dividing {a} by {b}")
    try:
        result = a/b
        logging.info("Division successful")
        return result

    except ZeroDivisionError:
        logging.error("Cannot divide by zero")  
        return None 
print(divide(10,2))   
print(divide(10,0)) 
```
