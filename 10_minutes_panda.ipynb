Creating a dataframe by passings a dictionary of objects where the keys are the column lables and the vlaues are the column values.

```python
df2 = pd.DataFrame(
    {
        "A":1.0,
        "B":pd.Timestamp("20130102"),
        "C":pd.Series(1, index=list(range(4)),dtype="float32"),
        "D":np.array([3]*4,dtype="int32"),
        "E":pd.Categorical(["test","train","test","train"]),
        "F":"foo",

    }
)

df2
```

Getitem([])

For a dataFrame, passing a single label selects a column and yields a series:

```python
df2["A"]
```

If the label only containes letter, numbers, and ubderscores, you can alternatively use the column name attribute:

```python
df2.A
```

Passing alist of column labels selects multiple columns, which can be useful for getting a subset/rearranging:

```python
df2[['B',"A"]]
```

For a DataFrame, passing a slice: selects matching row:

```python
df2[1:3]
```

Selection by label

see more in Selection by label using DataFrame.loc() or DataFrame.at().

Selecting a row matching a label:

```python
df2.loc[[0]]
```

Selecting all rows (:) with a select column labels

```python
df.loc[:,["A","B"]]
```

For label slicing, both endpoints are include:

```python
df.loc["20130103":"20130104",["A","B"]]
```

Selecting a single row and column label returns a scaler:

```python
df.loc[dates[0],"A"]
```

For getting fast access to a scaler(equivalent to the prior method)

```python
df.at[dates[0],"A"]
```

Select via the position of the passed integers:
