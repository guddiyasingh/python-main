1.Line Chart

Used to show trends overtime.

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,20,15,25,30]
plt.plot(x,y)
plt.title("Line Chart")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.show()
```

2. Bar Chart
Used to compare categories

```python
import matplotlib.pyplot as plt

students = ["Aman","Riya","Guddiya","Rahul"]
marks = [85,92,90,80]

plt.bar(students , marks)
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

3. Scatter Plot

Shows the relationship between two numerical variables

```python
import matplotlib.pyplot as plt

height = [150,160,165,170,175]
weight = [45,50,55,65,70]

plt.scatter(height,weight)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```

4.Histogram
shows the distribution of values.

```python
import matplotlib.pyplot as plt

marks = [60,65,70,75,80,85,90,95,100]

plt.hist(marks)
plt.title("Marks Distribution")
plt.show()
```

5. Pie Chart
Shows proportion

```python
import matplotlib.pyplot as plt

lables = ["IT","HR","Finance"]
students = [40,30,30]

plt.pie(students, labels=lables,autopct="%1.if%%")
plt.title("Department Distribution")
plt.show()
```

Matplotlib with Pandas
Suppose you have a DtaFrame:

```python
import pandas as pd

data={
    "Name":["Guddiya","Aman","Riya","Rahul"],
    "Marks":[90,85,95,80]
}
df=pd.DataFrame(data)
```

```python
import matplotlib.pyplot as plt

plt.bar(df["Name"], df["Marks"], color="black")
plt.title("Students Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

Customize Charts

Add color

```python
plt.bar(["Name"],df["Marks"],color="green")
```

Add grid:

```python
plt.grid(True)
```

Change line style:

```python
plt.plot(x,y, linestyle="--",marker="o")
```

Increase figure size:

```python
plt.figure(figsize=(8,5))
```
