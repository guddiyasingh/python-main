* Aggrigate functions

```python
import pandas as pd

data = {
    'Name':["Guddiya","Aman","Rahul","Riya"],
    'Marks':[90,85,95,80],
    'Age':[21,23,22,24]
}
df = pd.DataFrame(data)
print(df)
```

```python
# 1. sum
print(df['Marks'].sum())
```

```python
# 2.Mean Average
print(df["Marks"].mean())
```

```python
# 3.Maximum
print(df["Marks"].max())
```

```python
# 4.Minimum
print(df["Marks"].min())
```

```python
#5.count
print(df["Marks"].count())
```

```python
# 6. Standard Deviation
# measures how spread out the values are.
print(df["Marks"].std())
```

```python
# 7. Variance

print(df["Marks"].var())
```

```python
# 8.Median
# The middle value after sorting
print(df["Marks"].median())
```

```python
# 9.Mode
# The most frequently occuring value.
print(df["Marks"].mode())
```

```python
marks = [80,90,90,85]
print(pd.Series(marks).mode())
```

```python
# 10. Multiple aggrigret Function

print(df['Marks'].agg(['sum','mean','max','min']))

```

Agrrigrant functios with groupby()

```python
data = {
    "Department": ['IT','IT','HR','HR'],
    "Salary":[400000,450000,50000,4500]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Average Salary by depatment
print(df.groupby('Department')['Salary'].mean())
```

```python
#  Multiple Aggregate Functions
print(df.groupby('Department')['Salary'].agg(["mean","sum","max","min"]))
```
