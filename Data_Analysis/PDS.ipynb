A Series is a one_dimensional labled Array

```python
import pandas as pd
s = pd.Series([10,20,30,40,50])
print(s)
```

Creating a series with custom Index

```python
import pandas as pd

s=pd.Series([90,85,95], index = ["Math","Science","English"])
print(s)
```

Acess data:

```python
print(s["Math"])
```

A DataFram is a 2-dimensional table (rows and columns)

```python
# Create a DataFrame

import pandas as pd

data = {

   "Name":["Guddiya","Aman","Riya"],
   "Age":[22,23,21],
   "city":["Delhi","Mumbai","Pune"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Access a column

print(df["Name"])
```

```python
# Access rows

# Using.loc[]

print(df.loc[0])
```

```python
# Using .iloc[]
print(df.iloc[1])
```

```python
# Series
## A panda series is a 1_diemensional array like object that can hold any data 
import pandas as pd
 
data =[1,2,3,4,5] 
series = pd.Series(data)
print("Series\n",series)
print(type(series))
```

```python
## Create a series from dictionary
data = {'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data = [1,2,3,4,5]
index=['a','b','c','d','e']
pd.Series(data,index=index)
```

```python
# DataFrame
## create a dataFrame from a Dictionary

data={
    "Name":['krish','jhon','jack'],
     "Age":["20",'25','30'],
     "city":['Ranchi',"Delhi","pune"]
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```
