What is Seaborn

Seaborn is Python library built on top of Matplotlib. It makes it much easier to create beautiful, infomative statistical visualizations with just a few lines of code.

Matplotlib gives you flexiblity, while Seaborn gives you attractive defaults and simpler syntax.

```python
import seaborn as sns

df = sns.load_dataset("tips")

print(df.head())
```

1. Scatter Plot
Shows the relationship between two numerical variables.

```python
sns.scatterplot(data=df, x="total_bill",y="tip")
plt.title("Total Bill vs Tip")
plt.show()
```

```python
# 2. Line Plot
sns.lineplot(data=df, x="size",y="total_bill")
plt.show()
```

3. Bar PLot
Shows the average value for each category by default.

```python
sns.barplot(data=df, x="day",y="total_bill")
plt.show()
```

Count Plot
Counts hoe many observation belong t o each category.

```python
sns.countplot(data=df, x="day")
plt.show()
```

5. Histogram
Shows the distribution of a numerical column.

```python
sns.histplot(df["total_bill"])
plt.show()
```

Box Plot

Useful for finding outliers and understanding spread.

```python
sns.boxplot(data=df, x="day",y="total_bill")
plt.show()
```

7. Violin Plot
shows the distribution and density of the data

```python
sns.violinplot(data=df, x='day',y='total_bill')
plt.show()
```

8. Pair Plot
Compares every numerical column with every other numerical column.

```python
sns.pairplot(df)
plt.show()
```

9. Heatmap
Shows correlation between numerical columns

```python
corr =df.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.show()
```

10. Distribution Plot (Histograam+KDE)

```python
sns.histplot(df["tip"], kde=True)
plt.show()
```

Using Seaborn with Student Dataset

```python
import pandas as pd
students = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Department": ["IT", "HR", "IT", "Finance","HR"],
    "Marks": [90, 85, 95, 80, 92],
    "Age": [20, 22, 21, 23, 20]
}
df = pd.DataFrame(students)
```

```python
# Bar PLot
sns.barplot(data=df, x="Name",y="Marks")
plt.show()
```

```python
# Count Plot
sns.countplot(data=df, x="Department")
plt.show()
```

```python
# Box Plot
sns.boxplot(data=df, y="Marks")
plt.show()
```

```python
# Histogram
sns.histplot(df["Marks"], kde=True)
plt.show()
```

```python
# Scatter Plot
sns.scatterplot(data=df, x="Age",y="Marks")
plt.show()
```

```python
### Basic Plotting with Seaborn
tips = sns.load_dataset("tips")
tips
```

```python
## create a scatter plot
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python
sns.scatterplot(x='tip', y='total_bill', data=tips)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python
## Categorical Plots
## Box plot
sns.boxplot(x='day', y='total_bill', data=tips)
```

```python
## Violin plot
sns.violinplot(x='day', y='total_bill', data=tips)
```

```python
### Histogram 
sns.histplot(tips['total_bill'], bins=10,kde=True)
```

```python
## KDE Plot
sns.kdeplot(tips['total_bill'], shade=True)
```

```python
# PairPLot
sns.pairplot(tips)
```

```python
# Heatmap
corr = tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True,cmap='coolwarm', fmt=".2f", linewidths=0.7)
```
