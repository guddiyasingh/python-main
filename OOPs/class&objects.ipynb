```python

class Dog:
    # The Constructor
    def __init__(self, name, breed):
        self.name = name    # Attribute
        self.breed = breed  # Attribute
        
    # A Method (a function inside a class)
    def bark(self):
        return f"{self.name} says Woof!"

# Creating Objects (Instances of the class)
dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Max", "German Shepherd")

# print(dog1.bark())  # Outputs: Buddy says Woof!

print(dog1.breed)
print(dog2.bark())   # Outputs: German Shepherd
```

The 4 pillars of OOP

To truly master OOP, yu need to understand its four foundational concepts:

1. INheritance allows a new class (Child) to adopt attributes and methods from an existing class (Parent). This prevents you from repeating yurself

```python
# Parent Class
class Animal:

    def __init__(self,name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."

# Child Class inherits from Animal
class Cat(Animal):
    def meow(self):
        return f"{self.name} says Meow"     
my_cat = Cat("Whiskers")   
print(my_cat.eat()) # INherited method: Whiskers is eating.
print(my_cat.meow()) #Child method: Whiskers says Meow! 
```

```python
# The Blueprint (Class)
class Car:
    def __init__(self, color, model):
        # Attributes: What the car HAS
        self.color = color
        self.model = model
        self.odometer = 0

    # Method: What the car DOES
    def drive(self, miles):
        self.odometer += miles
        return f"The {self.color} {self.model} drove {miles} miles."

# The Actual Cars (Objects)
my_car = Car("Red", "Tesla Model 3")
your_car = Car("Matte Black", "Ford Mustang")

print(my_car.drive(50))     # Outputs: The Red Tesla Model 3 drove 50 miles.
print(your_car.odometer)    # Outputs: 0 (Because your car hasn't driven yet!)
```

```python
#Read a text file and count the number of lines , words, and characters.
# Counting lines, words, and characters in a text file

def count_text_file(file_path):
    with open(file_path,'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'example.txt'
lines , words, characters = count_text_file(file_path)
print(f'Lines:{lines},Words: {words},Characters: {characters}')    
```

```python
class student:
    def __init__(self, name,marks):
        self.name = name
        self.marks=marks

    def get_avg(self):
        sum=0
        for val in self.marks:
            sum+=val
        print("hi",self.name,"your average score:",sum/3) 

s=student("karan",[12,13,24])  
print(s.name,s.marks)   
s.get_avg()            
```

```python
class staff:
    def __init__(self, name, shift):
        self.name = name
        self.shift = shift

    def start_work(self):
        print(f"{self.name} starts work on {self.shift} shift")

class waitor(staff):
    def take_order(self):
        print(f"{self.name} is taking an orders")

class chef(staff):
    def cook_food(self):
        print(f"{self.name} is cooking food")

raj = waitor("Raj","Morning")
Amit=chef("Amit", "Night")

raj.waitor()
Amit.chef()
```
