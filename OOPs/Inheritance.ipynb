Inheritance In python

Inheritance is a fundamental concept in object-orientend programming(oop) that allows a class to inherit attribute and methods from another class. This lesson covers single inheritance and multiple inhritance, demonstrating how to create and use them in python.

```python
# Inheritance
# Parent class

class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
    def drive(self):
        print(f"The person will drive the {self.enginetype} car")     
```

```python
car1=Car(4,5,"petrol")
print(car1.drive())
```
