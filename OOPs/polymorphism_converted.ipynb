# Polymorphism in Python
This notebook demonstrates polymorphism via method overriding and duck typing.

- `Animal` is the base class with a `speak()` interface.
- `Dog`, `Cat`, and `Bird` override `speak()`.
- A `Robot` shows duck typing (any object with `speak()` works).

```python
class Animal:
    """Base class with a common interface."""

    def speak(self):
        raise NotImplementedError("Subclasses must implement speak()")

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

class Bird(Animal):
    def speak(self):
        return "Tweet!"

def animal_sounds(animals):
    """Demonstrates polymorphism by calling the same method on different types."""
    for animal in animals:
        print(f"{animal.__class__.__name__}: {animal.speak()}")
```

```python
# Create instances and run the demonstration
pets = [Dog(), Cat(), Bird()]
animal_sounds(pets)

# Duck typing example: any object with a `speak` method works.
class Robot:
    def speak(self):
        return "Beep boop"

print("-- Duck typing example --")
animal_sounds(pets + [Robot()])
```

```python
# Inheritance example added: single and multiple inheritance
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hello, I'm {self.name}"

class Student(Person):
    def __init__(self, name, student_id):
        super().__init__(name)
        self.student_id = student_id

    def greet(self):
        return f"{super().greet()} and I'm a student (ID: {self.student_id})"

class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject

    def greet(self):
        return f"{super().greet()} and I teach {self.subject}"

class Researcher:
    def research(self):
        return "Doing research"

class Professor(Teacher, Researcher):
    pass

people = [Person("Alex"), Student("Bella", 123), Teacher("Carlos", "Math"), Professor("Dana", "Physics")]
for p in people:
    print(f"{p.__class__.__name__}: {p.greet()}")

prof = Professor("Eve", "Computer Science")
print(f"Professor research: {prof.research()}")
```
