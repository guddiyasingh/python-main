```python
## if statment
age=10
if age < 18:
    print("You are a child")
```

```python
## else
## this else statment will execute when the condition in if statment is false.

age = 20
if age >= 18:
    print("You are eligible for voting")
else:
    print("You are minor")    
```

```python
## elif
## The elif statment allows you to check multiple condotions . It stands for else iffor "

age =20

if age<13:
    print("You are a child")
elif age < 20:
    print("You are a teenager")    
else:
    print("you are an adult")    
```
