Step4: Create  a Table

```python
cursor.execute("""

CREATE TABLE students(
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER
)
""")

# <!-- SAVE THE CHANGES -->
conn.commit()
```

Step 5 : Insert Data

```python
cursor.execute("""
INSERT INTO students(name, age, marks)
VALUES('johan',20,90)
""")
conn.commit()
```

```python
# Insert multiple rows

students =[
    ("Aman", 23,90),
    ("Riya",21,98),
    ("Rahul",22,85)
    ]

cursor.executemany(
    "INSERT INTO students(name, age, marks) VALUEs(?,?,?)",
    students
    )
conn.commit()
```

```python
import sqlite3

# Connect to database
conn = sqlite3.connect("students.db")

#Create a cursor
cursor = conn.cursor()

#Create table
cursor.execute(""" 

CREATE TABLE IF NOT EXISTS students(
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER)""")

# Insert data into table
cursor.execute("""
INSERT INTO students(name, age, marks)
VALUES('john', 20, 90)
""")
# Save changes
conn.commit()

# Read Data
cursor.execute("SELECT * FROM students")
for row in cursor.fetchall():
    print(row)

# Close the connection
conn.close()
```

Using Pandas with SQLite
you can load SQL query results directly into a Pandas DataFrame.

```python
import sqlite3
import pandas as pd

conn=sqlite3.connect("student.db")
df=pd.read_sql_query("SELECT * FROM students", conn)

print(df)
conn.close()
```

```python
## Working with sales data
# connect to an SQLite database
connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()

# Create a table for sales data
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales (
id INTEGER PRIMARY KEY,
date TEXT NOT NULL,
product TEXT NOT NULL,
sales INTEGER,
region TEXT
)
''')
# Insert data into sales table

sales_data = [
  ('2023-01-01','Product', 100,'North'),
  ('2023-01-02','Product', 100,'South'),
  ('2023-01-03','Product', 100,'East'),
  ('2023-01-04','Product', 100,'West'),
  ('2023-01-04','Product', 100,'North')

]
cursor.executemany('''
Insert into sales (date,product,sales,region)
                   values(?,?,?,?)    

''',sales_data)
connection.commit()
```

```python
# Query data from the sales table
cursor.execute('SELECT * FROM sales')
rows = cursor.fetchall()

# Print the queried data

for row in rows:
    print(row)
```
