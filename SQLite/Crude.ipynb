step2:Create or Connect to a Database

```python
import sqlite3
conn=sqlite3.connect("student.db")
print("Database connected!")

# if student.db does't exist , SQLite creates it automaticaly.
```

Step3:Create a Cursor
A cursor lets you execute SQL commands.

```python
cursor  = conn.cursor()
```

Step4: Create  a Table

```python
cursor.execute("""

CREATE TABLE students(
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER
)
""")

# <!-- SAVE THE CHANGES -->
conn.commit()
```

Step 5 : Insert Data

```python
cursor.execute("""
INSERT INTO students(name, age, marks)
VALUES('johan',20,90)
""")
conn.commit()
```

```python
# Insert multiple rows

students =[
    ("Aman", 23,90),
    ("Riya",21,98),
    ("Rahul",22,85)
    ]

cursor.executemany(
    "INSERT INTO students(name, age, marks) VALUEs(?,?,?)",
    students
    )
conn.commit()
```

step 6 :Read Data

```python
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

step 7:Read One Row


```python
cursor.execute("SELECT * FROM students")
print(cursor.fetchone())
```

step8: Filter Data

```python
cursor.execute("""
SELECT * FROM students
WHERE marks > 90
""")
print(cursor.fetchall())
```

step 9: UPdate Data

```python
cursor.execute("""
UPDATE students
SET marks = 99
WHERE name = 'Guddiya'
""")
conn.commit()
# print(cursor.fetchall())
```

```python
import sqlite3

# Connect to database
conn = sqlite3.connect("students.db")

#Create a cursor
cursor = conn.cursor()

#Create table
cursor.execute(""" 

CREATE TABLE IF NOT EXISTS students(
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER)""")

# Insert data into table
cursor.execute("""
INSERT INTO students(name, age, marks)
VALUES('john', 20, 90)
""")
# Save changes
conn.commit()

# Read Data
cursor.execute("SELECT * FROM students")
for row in cursor.fetchall():
    print(row)

# Close the connection
conn.close()
```

Using Pandas with SQLite
you can load SQL query results directly into a Pandas DataFrame.

```python
import sqlite3
import pandas as pd

conn=sqlite3.connect("student.db")
df=pd.read_sql_query("SELECT * FROM students", conn)

print(df)
conn.close()
```
